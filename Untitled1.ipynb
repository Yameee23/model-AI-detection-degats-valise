```python
from PIL import Image
img = Image.open("/content/Bag Damage Detection Dataset/test/images/steptodown-com696734_jpg.rf.85ca1fb0e0f5801550558512e35b4671.jpg")
largeur, hauteur = img.size
print(largeur, hauteur)
```

```python
with open("/content/model-AI-detection-degats-valise/Bag Damage Detection Dataset/test/labels/steptodown-com696734_jpg.rf.85ca1fb0e0f5801550558512e35b4671.txt") as f:
    contenu = f.read()
print(contenu)
```

```python
nombres = [float(x) for x in contenu.split()]
print(nombres)
```

```python
coord= nombres[1:]
print(coord)
```

```python
points=[(coord[i],coord[i+1])for i in range(0,len(coord),2)]
print(point)
```

```python
points_pixels =[(x * largeur, y * hauteur) for x, y in points]
print(points_pixels)
```

```python
import matplotlib.pyplot as plt

plt.imshow(img)
polygone = plt.Polygon(points_pixels,fill=False,color="red")
plt.gca().add_patch(polygone)
plt.show()

```
